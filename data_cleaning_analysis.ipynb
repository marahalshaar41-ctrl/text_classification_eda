```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("df_file.csv")
```

```python
import re

# Convert text to lowercase
df["clean_text"] = df["Text"].str.lower()

# Remove punctuation
df["clean_text"] = df["clean_text"].str.replace(r"[^\w\s]", "", regex=True)

df[["Text", "clean_text"]].head()
```

```python
# Split text into words
df["tokens"] = df["clean_text"].str.split()
df[["clean_text", "tokens"]].head()
```

```python
import nltk

# Download stopwords
#nltk.download("stopwords")

# Get English stopwords
#stop_words = nltk.corpus.stopwords.words("english")
```

```python
# Remove stopwords
tokens_no_stopwords = []

for words in df["tokens"]:
    clean_words = []
    for word in words:
        if word not in stop_words:
            clean_words.append(word)
    tokens_no_stopwords.append(clean_words)

df["tokens_no_stopwords"] = tokens_no_stopwords

# Show the result
df[["tokens", "tokens_no_stopwords"]].head()
```

```python
df["clean_text"] = df["clean_text"].str.strip()# Remove extra spaces
```

```python
# Join words into a single text
final_text = []

for words in df["tokens_no_stopwords"]:
    text = " ".join(words)
    final_text.append(text)

df["final_text"] = final_text

# Show the result
df[["Text", "final_text"]].head()
```
